## Prompt untuk menyusun recipe
Recipe adalah instruksi kerja yang dapat digunakan kembali untuk menyelesaikan satu problem
yang cukup sempit, dari kondisi awal hingga hasil yang dapat diperiksa. Dalam buku ini recipe
bukan sekadar contoh prompt, potongan code, atau daftar command. Ia menghubungkan tujuan
ilmiah, input, assumptions, procedure, artifact, serta pemeriksaan yang diperlukan agar pembaca
dapat memahami apa yang dikerjakan dan mengulangi bagian yang relevan.

```python
import numpy as np

# 1. Data curah hujan
tanggal = [
    "2026-01-01",
    "2026-01-02",
    "2026-01-03",
    "2026-01-04",
    "2026-01-05",
    "2026-01-06",
    "2026-01-07"
]

curah_hujan = np.array([
    0, 12.5, np.nan, 8, 0, 25, 4
])

# 2. Memeriksa data
valid = curah_hujan[~np.isnan(curah_hujan)]

jumlah_data = len(curah_hujan)
jumlah_valid = len(valid)
jumlah_hilang = np.isnan(curah_hujan).sum()

# 3. Menghitung statistik
rata_rata = np.mean(valid)
maksimum = np.max(valid)
hari_hujan = np.sum(valid > 0)

# 4. Menampilkan data
print("DATA CURAH HUJAN")
for tgl, hujan in zip(tanggal, curah_hujan):
    print(f"{tgl}: {hujan} mm")

# 5. Menampilkan hasil analisis
print("\nHASIL ANALISIS CURAH HUJAN")
print(f"Jumlah seluruh data  : {jumlah_data}")
print(f"Data valid           : {jumlah_valid}")
print(f"Data hilang          : {jumlah_hilang}")
print(f"Rata-rata curah hujan: {rata_rata:.2f} mm")
print(f"Curah hujan maksimum : {maksimum:.2f} mm")
print(f"Jumlah hari hujan    : {hari_hujan}")

# 6. Menyimpan hasil
with open("hasil_analisis.txt", "w", encoding="utf-8") as file:
    file.write("HASIL ANALISIS CURAH HUJAN\n")
    file.write(f"Jumlah seluruh data  : {jumlah_data}\n")
    file.write(f"Data valid           : {jumlah_valid}\n")
    file.write(f"Data hilang          : {jumlah_hilang}\n")
    file.write(f"Rata-rata            : {rata_rata:.2f} mm\n")
    file.write(f"Maksimum             : {maksimum:.2f} mm\n")
    file.write(f"Jumlah hari hujan    : {hari_hujan}\n")

print("\nHasil tersimpan di hasil_analisis.txt")
```
