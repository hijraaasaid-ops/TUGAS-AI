**Data:**
[Masukkan gambar seismogram atau tabel data di sini]


```python
import pandas as pd
import matplotlib.pyplot as plt

# =========================
# DATA SEISMIK
# =========================

data = {
    "Stasiun": ["GRFO", "GNI", "OBN", "KONO", "KEV"],
    "Tp": [14782, 14733, 14782, 14821, 14897],
    "Ts": [15028, 14933, 15028, 15095, 15230]
}

df = pd.DataFrame(data)

# =========================
# PERHITUNGAN S-P
# =========================

df["S-P"] = df["Ts"] - df["Tp"]

# =========================
# MENAMPILKAN HASIL
# =========================

print("HASIL PICKING GELOMBANG P DAN S")
print(df)

# =========================
# VALIDASI
# =========================

if (df["Tp"] < df["Ts"]).all():
    print("\nSemua data memenuhi Tp < Ts")
else:
    print("\nTerdapat data yang tidak memenuhi Tp < Ts")

# =========================
# GRAFIK
# =========================

plt.figure(figsize=(9, 5))

plt.plot(df["Stasiun"], df["Tp"],
         marker="o", label="Gelombang P")

plt.plot(df["Stasiun"], df["Ts"],
         marker="o", label="Gelombang S")

plt.xlabel("Stasiun")
plt.ylabel("Waktu Tiba (s)")
plt.title("Waktu Tiba Gelombang P dan S")
plt.grid(True)
plt.legend()

plt.show()
```

Kode ini digunakan untuk melihat waktu datang gelombang P dan S pada beberapa stasiun.

Misalnya kode menghitung:

delta_t = Ts - Tp
Maksud hasilnya

Tp = waktu tiba gelombang P

Ts = waktu tiba gelombang S

Ts - Tp = selisih waktu kedatangan gelombang P dan S.

Contohnya:

Tp = 100 s
Ts = 160 s

Ts - Tp = 60 s

Artinya gelombang S tiba 60 detik setelah gelombang P.

Semakin besar nilai Ts - Tp, biasanya menunjukkan bahwa jarak stasiun terhadap sumber gempa semakin jauh, karena gelombang P bergerak lebih cepat daripada gelombang S.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =========================
# DATA
# =========================

data = {
    "Stasiun": ["GRFO", "GNI", "OBN", "KONO", "KEV"],
    "Tp": [14782, 14733, 14782, 14821, 14897],
    "Ts": [15028, 14933, 15028, 15095, 15230]
}

df = pd.DataFrame(data)

# =========================
# HITUNG S-P
# =========================

df["S-P"] = df["Ts"] - df["Tp"]

print("Tabel Perhitungan Diagram Wadati")
print(df)

# =========================
# REGRESI LINEAR
# =========================

x = df["Tp"]
y = df["Ts"]

koef = np.polyfit(x, y, 1)

m = koef[0]
b = koef[1]

# Prediksi
y_pred = m * x + b

# R^2
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)

r2 = 1 - (ss_res / ss_tot)

# =========================
# Vp/Vs
# =========================

Vp_Vs = m

# =========================
# HASIL
# =========================

print("\nPersamaan regresi:")
print(f"Ts = {m:.4f} Tp + {b:.4f}")

print(f"\nKemiringan = {m:.4f}")
print(f"R² = {r2:.4f}")
print(f"Vp/Vs = {Vp_Vs:.4f}")

# =========================
# GRAFIK WADATI
# =========================

plt.figure(figsize=(9, 6))

plt.scatter(x, y, label="Data Stasiun")

plt.plot(
    x,
    y_pred,
    label=f"Regresi: Ts = {m:.4f}Tp + {b:.4f}"
)

plt.xlabel("Tp (s)")
plt.ylabel("Ts (s)")
plt.title("Diagram Wadati")
plt.grid(True)
plt.legend()

plt.show()
```

Kode di atas digunakan untuk membuat Diagram Wadati dan menentukan hubungan antara waktu kedatangan P dan S.

Biasanya kode menghitung:

Ts_Tp = Ts - Tp

kemudian melakukan regresi linear:

model = LinearRegression()

dan menghasilkan:

slope/kemiringan
persamaan garis
R²
perkiraan Vp/Vs
origin time T0
a. Slope

Misalnya hasilnya:

slope = 0.78

Nilai slope digunakan untuk memperoleh perbandingan

Artinya kecepatan gelombang P diperkirakan sekitar 1,78 kali kecepatan gelombang S.

b. R²

Misalnya:

R² = 0.99

R² menunjukkan seberapa baik titik-titik data mengikuti garis regresi.

Nilai mendekati 1 berarti hubungan datanya sangat linear.

Jadi:

R² yang tinggi menunjukkan bahwa hubungan antara waktu kedatangan P dan S pada data cukup konsisten sehingga regresi linear dapat digunakan dengan baik.

c. Origin time / T₀

Kode juga dapat memperkirakan:

T0 = waktu terjadinya gempa

Ini berbeda dengan Tp.

Tp = gempa sampai di stasiun.

T0 = perkiraan waktu ketika gempa mulai terjadi di sumbernya.

Jadi Diagram Wadati dapat digunakan untuk memperkirakan waktu asal gempa (origin time).
